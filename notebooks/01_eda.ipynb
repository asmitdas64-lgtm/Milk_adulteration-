# Phase 2 — Exploratory Data Analysis (EDA)
### Machine Learning for Predicting Food Adulteration
**Dataset:** Milk Quality & Adulteration Dataset (`data/interim/clean.csv`)

```python
from pathlib import Path
import pandas as pd
from foodadult.config import load_config
from foodadult.eda import generate_summary_stats, compute_iqr_outliers

cfg = load_config("../configs/default.yaml")
df = pd.read_csv("../data/interim/clean.csv")
df.head()
```

## 1. Class Distribution
0 = Pure / High Quality  
1 = Adulterated / Substandard

```python
summary = generate_summary_stats(df, target_col="label")
print("Row count:", summary["row_count"])
print("Class distribution:", summary["class_balance"]["counts"])
print("Proportions:", summary["class_balance"]["proportions"])
```

## 2. IQR Outlier Analysis
Examining extreme physicochemical deviations (spoilage / neutralizers / water dilution).

```python
outliers = compute_iqr_outliers(df, cfg["data"]["features"], target_col="label")
for feat, info in outliers.items():
    print(f"{feat}: {info['total_outliers']} outliers | Bounds: [{info['lower_bound']:.2f}, {info['upper_bound']:.2f}]")
```
